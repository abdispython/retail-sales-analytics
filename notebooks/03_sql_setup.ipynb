```python
import pandas as pd
import sqlite3
from pathlib import Path
```

```python
project_folder = Path.cwd().parent

csv_path = project_folder / "data" / "train.csv"
database_path = project_folder / "data" / "retail_sales.db"

print("CSV:", csv_path)
print("Database:", database_path)
```

```python
df = pd.read_csv(csv_path)

df.head()
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("-", "_")
)

df.columns.tolist()
```

```python
df.info()
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], dayfirst=True)
df["ship_date"] = pd.to_datetime(df["ship_date"], dayfirst=True)

df[["order_date", "ship_date"]].head()
```

```python
df["postal_code"] = df["postal_code"].astype("Int64")

df[["postal_code"]].head()
```

```python
df.info()
```

```python
df.head()
```

```python
connection = sqlite3.connect(database_path)

print("Connected to SQLite database.")
```

```python
df.to_sql(
    name="retail_sales",
    con=connection,
    if_exists="replace",
    index=False
)

print("Data loaded into the retail_sales table.")
```

```python
tables_query = """
SELECT name
FROM sqlite_master
WHERE type = 'table';
"""

pd.read_sql_query(tables_query, connection)
```

```python
preview_query = """
SELECT *
FROM retail_sales
LIMIT 5;
"""

pd.read_sql_query(preview_query, connection)
```

```python
row_count_query = """
SELECT COUNT(*) AS total_rows
FROM retail_sales;
"""

pd.read_sql_query(row_count_query, connection)
```

```python
schema_query = """ 
PRAGMA table_info(retail_sales);
"""

pd.read_sql_query(schema_query, connection)
```

```python
query = """ 
SELECT 
    ROUND(SUM(sales), 2) AS total_sales
FROM retail_sales;
"""

pd.read_sql_query(query, connection)
```

```python
category = """ 
SELECT
    category,          
    ROUND(SUM(sales), 2) AS total_category_sales         
FROM retail_sales
GROUP BY
    category
ORDER BY
    total_category_sales desc;
""" 

pd.read_sql_query(category, connection)
```

```python
product = """ 
SELECT

    product_name,

    ROUND(SUM(sales), 2) AS total_product_sales

FROM retail_sales

GROUP BY

    product_name

ORDER BY

    total_product_sales DESC
LIMIT 10;
"""

pd.read_sql_query(product, connection)
```

```python
region = """ 
SELECT
    ROUND(SUM(sales), 2) AS western_sales
FROM retail_sales
WHERE
    region = 'West';
"""

pd.read_sql_query(region, connection)
```

```python
region = """ 
SELECT
    category,
    ROUND(SUM(sales), 2) AS western_sales
FROM retail_sales
WHERE
    region = 'West'
GROUP BY
    category
ORDER BY
    western_sales DESC;
"""

pd.read_sql_query(region, connection)
```

```python
west_customers = """ 
SELECT
    customer_name,
    ROUND(SUM(sales), 2) AS west_customers
FROM retail_sales
WHERE
    region = 'West'
GROUP BY
    customer_name
ORDER BY
    west_customers DESC
LIMIT 10;
"""

pd.read_sql_query(west_customers, connection)
```

```python
orders = """ 
SELECT
    region,
    COUNT(order_id) AS total_orders
FROM retail_sales
GROUP BY
    region
ORDER BY
    total_orders DESC;
"""

pd.read_sql_query(orders,connection)
```

```python
customer_orders = """ 
SELECT
    customer_name,
    COUNT(order_id) AS total_orders
FROM retail_sales
GROUP BY
    customer_name
ORDER BY
    total_orders DESC
LIMIT 10;
"""

pd.read_sql_query(customer_orders, connection)
```

```python
average_sales = """ 
SELECT
    category,
    ROUND(AVG(sales), 2) AS average_sales
FROM retail_sales
GROUP BY category
ORDER BY average_sales DESC;
"""

pd.read_sql_query(average_sales, connection)
```

```python
total_sales = """ 
SELECT
    category,
    ROUND(SUM(sales), 2) AS total_sales
FROM retail_sales
GROUP BY category
HAVING SUM(sales) > 700000
ORDER BY total_sales DESC;
"""

pd.read_sql_query(total_sales,connection)
```

```python
max_min_sales = """ 
SELECT
    category,
    ROUND(MAX(sales), 2) AS highest_sale,
    ROUND(MIN(sales), 2) AS lowest_sale
FROM retail_sales
GROUP BY category
ORDER BY highest_sale DESC;
"""

pd.read_sql_query(max_min_sales,connection)
```

```python
technology_sales = """ 
SELECT
    category,
    ROUND(sales, 2) AS sales
FROM retail_sales
WHERE region = 'West'
  AND category = 'Technology'
  AND sales > 500
ORDER BY sales DESC
LIMIT 10;
"""

pd.read_sql_query(technology_sales,connection)
```

```python
region_sales = """ 
SELECT
    region,
    category,
    ROUND(sales, 2) AS sales
FROM retail_sales
WHERE
    (region = 'East' OR region = 'West')
    AND sales > 1000
ORDER BY sales DESC
LIMIT 10;
"""

pd.read_sql_query(region_sales,connection)
```

```python
chair_sales = """ 
SELECT
    product_name,
    category,
    ROUND(sales, 2) AS sales
FROM retail_sales
WHERE product_name LIKE '%Chair%'
ORDER BY sales DESC
LIMIT 10;
"""

pd.read_sql_query(chair_sales,connection)
```

```python
furniture_technology_sales = """ 
SELECT
    product_name,
    category,
    ROUND(sales, 2) AS sales
FROM retail_sales
WHERE category IN ('Furniture', 'Technology')
  AND sales > 500
ORDER BY sales DESC
LIMIT 10;
"""

pd.read_sql_query(furniture_technology_sales,connection)
```

```python
between_sales = """ 
SELECT
    order_id,
    product_name,
    ROUND(sales, 2) AS sales
FROM retail_sales
WHERE sales BETWEEN 500 AND 1000
ORDER BY sales DESC
LIMIT 10;
"""

pd.read_sql_query(between_sales,connection)
```

```python
case_when = """ 
SELECT
    product_name,
    ROUND(sales, 2) AS sales,
    CASE
        WHEN sales < 100 THEN 'Low'
        WHEN sales BETWEEN 100 AND 500 THEN 'Medium'
        ELSE 'High'
    END AS sales_category
FROM retail_sales
LIMIT 10;
"""

pd.read_sql_query(case_when,connection)
```
